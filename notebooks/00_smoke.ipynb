```python
import pandas as pd
from IPython.display import display
from tcco2_accuracy.workflows import bootstrap, infer, meta, paco2, sim
```

```python
meta_result = meta.run_meta_checks()
display(meta_result.summary)
print(meta_result.invariants)
```

```python
bootstrap_result = bootstrap.run_bootstrap(n_boot=200, seed=202401)
display(bootstrap_result.summary)
print(bootstrap_result.invariants)
```

```python
synthetic_paco2 = pd.DataFrame({
    'paco2': [35.0, 45.0, 55.0, 30.0, 45.0, 60.0, 25.0, 45.0, 65.0],
    'subgroup': ['pft'] * 3 + ['ed_inp'] * 3 + ['icu'] * 3,
})
paco2_result = paco2.run_paco2_summary(paco2_data=synthetic_paco2)
display(paco2_result.summary)
print(paco2_result.invariants)
```

```python
simulation_result = sim.run_forward_simulation_summary(
    params=bootstrap_result.draws,
    paco2_data=paco2_result.data,
    seed=202401,
    n_draws=200,
)
display(simulation_result.summary)
print(simulation_result.invariants)
```

```python
inference_result = infer.run_inference_demo(
    tcco2_values=[35, 45, 55],
    params=bootstrap_result.draws,
    paco2_data=paco2_result.data,
    seed=202401,
    n_draws=200,
)
display(inference_result.summary)
print(inference_result.invariants)
```
